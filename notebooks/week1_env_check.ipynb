```python
import torch

# 模拟 5 个患者、48 小时、6 项体征的监护数据
data = torch.randn(5, 48, 6)
print("数据形状:", data.shape)

# 任务1：抽出所有患者的心率（假设心率是第0项体征）
hr = data[:, :, 0]
print("心率形状:", hr.shape)

# 任务2：取每个患者的最后6小时
last_6h = data[:, -6:, :]
print("最后6小时形状:", last_6h.shape)

# 任务3：找出所有"心率>1"的时点有多少个（模拟筛选高心率时点）
high_hr_count = (hr > 1).sum()
print("高心率时点数量:", high_hr_count)
```
